```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("data.csv")
```

```python
df.info()
```

```python
df["age"] = df["age"].apply(lambda x: str(x)).apply(lambda x: x.strip())
```

```python
df.loc[df["age"] != "NA", "age"].median()
```

```python
df.loc[df["age"] == "NA", "age"] = "39"
```

```python
df["age"] = df["age"].astype(int)
```

```python
df.fecha_dato.unique()
```

```python
import seaborn as sns
```

```python
#Alterar o estilo do gráfico para o estilo padrão do seaborn
sns.set_style("darkgrid")
```

```python
#plotting fecha_dato vs. count and rotate x-axis labels by 45 degrees and gray colormap
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

plt.figure(figsize=(12,6))
sns.countplot(x="fecha_dato", data=df, color="red")
plt.ylabel('Quantidade', fontsize=12)
plt.xlabel('Data', fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

```python
#plot histplot of age
plt.figure(figsize=(12,6))
sns.distplot(df.age.values, bins=160, kde=False, color="black")
plt.xlabel('Idade', fontsize=12)
plt.show()
```

```python
#Faça um gráfico que mostre a relação entre segmento e age
plt.figure(figsize=(12,6))
sns.boxplot(x="segmento", y="age", data=df, color="gray")
plt.ylabel('Idade', fontsize=12)
plt.xlabel('Segmento', fontsize=12)
plt.show()
```

```python
df_a = df.loc[:, ['sexo', 'ind_actividad_cliente']].join(df.loc[:, "ind_ahor_fin_ult1": "ind_recibo_ult1"])
df_a = df_a.groupby(['sexo', 'ind_actividad_cliente']).sum()
df_a = df_a.T
```

```python
df_a.head()
```

```python
df_a.plot(kind='barh', stacked=True, fontsize=12, figsize=[12,6], colormap='Reds')
#plt.title('Popularidade de produtos por sexo e indice ativo', fontsize=12) 
plt.xlabel('Numero de Usuários', fontsize=12) 
plt.ylabel('Produtos', fontsize=12) 
plt.legend(["Sex:H; Activity_Ind:0", "Sex:H; Activity_Ind:1", "Sex:M; Activity_Ind:0", 
            "Sex:M; Activity_Ind:1"], prop={'size':12})
plt.show()
```
